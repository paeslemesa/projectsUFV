# Downloading GEDI raster from a list of files

```python
shp_dir = Path(r"D:\Pesquisa\PhD\IGARS2025\DatasetBiomes\biome_border\split")

out_dir = Path(r"G:\My Drive")


# Defining the dates for the analysis
start_date = '2019-03-01' # YYYY-MM-DD
end_date   = '2023-03-06' # YYYY-MM-DD
```

```python
shp_list = list(shp_dir.glob('*.geojson'))
print('Length of the list:', len(shp_list))
```

```python
for shp in tqdm(shp_list):
    
    if shp.stem == 'id_57' or shp.stem == 'id_56':
        continue
    
    print(shp.stem)
    ee_geom = dlGEE.shapefile_to_ee_geometry(shp)

    year_start = int(start_date[:4]) # Year of the start date
    year_end = int(end_date[:4]) # Year of the end date plus one (to include the last year)

    #print(f"Year start: {year_start}\nYear end: {year_end}")
    output_dir = f"GEE_exports_{shp.stem}"
    
    # test if out_dir exists, if not continue
    if Path(output_dir).exists():
        print(f"{output_dir} already exists. Skipping...")
        continue
    

    for year in range(year_start, year_end):
        date_ini = f"{year}-03-15"
        date_end = f"{year+1}-03-15"
        
        dlGEE.gedi_download(date_ini, date_end, ee_geom, output_dir).export_all_images()
```
